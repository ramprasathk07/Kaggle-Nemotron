```python
from mistralai import Mistral
import os

with Mistral(
    api_key=os.environ.get("MISTRAL_API_KEY", ""),
) as mistral:

    res = mistral.chat.complete(model=os.environ.get("MISTRAL_MODEL_NAME", ""), messages=[
        {
            "content": "Who is the best French painter? Answer in one short sentence.",
            "role": "user",
        },
    ], stream=False)

    # Handle response
    print(res)

```
