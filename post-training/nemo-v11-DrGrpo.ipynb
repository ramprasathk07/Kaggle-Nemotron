## Mode A: Train on Kaggle — Dr. GRPO

```python
if TRAIN_ON_KAGGLE:
    import os, gc, time, re, subprocess
    import pandas as pd
    import torch
    from datasets import Dataset as HFDataset
    from transformers import TrainerCallback

    os.environ["PYTORCH_ALLOC_CONF"] = "expandable_segments:True"
    os.environ["CUDA_LAUNCH_BLOCKING"] = "1"

    SEED = 42
    PROMPT_SUFFIX = '\nPlease put your final answer inside `\\boxed{}`. For example: `\\boxed{your answer}`'

    # ---- Dataset path: use competition problems (prompt+answer, no CoT needed for GRPO) ----
    # Option A — original competition split
    DATASET_PATH = "/kaggle/input/datasets/dgxchen/nemotron-cot-tong/problem_ids_matched.csv"
    # Option B — balanced merged dataset (uncomment if uploaded)
    # DATASET_PATH = "/kaggle/input/datasets/your-dataset/merged_cot_final.csv"

    df = pd.read_csv(DATASET_PATH)
    print(f"Dataset: {len(df)} rows")
    train_df = df.sample(frac=1, random_state=SEED).reset_index(drop=True)

    records, skipped = [], 0
    for _, row in train_df.iterrows():
        prompt = str(row["prompt"])
        answer = str(row["answer"])
        if not answer or answer == "nan":
            skipped += 1
            continue
        user_content = prompt + PROMPT_SUFFIX
        try:
            prompt_text = tokenizer.apply_chat_template(
                [{"role": "user", "content": user_content}],
                tokenize=False, add_generation_prompt=True, enable_thinking=True,
            )
        except TypeError:
            prompt_text = tokenizer.apply_chat_template(
                [{"role": "user", "content": user_content}],
                tokenize=False, add_generation_prompt=True,
            )
        records.append({"prompt": prompt_text, "answer": answer})

    grpo_dataset = HFDataset.from_list(records)
    print(f"GRPO dataset: {len(records)} records (skipped {skipped})")

    # ---- GPU metrics callback (same as v9-1) ----
    class GPUMetricsCallback(TrainerCallback):
        def __init__(self, log_every_n_steps=2):
            super().__init__()
            self.log_every_n_steps = log_every_n_steps
            self._last_step_time = None
            self._last_global_step = 0

        def _query_nvidia_smi(self):
            try:
                r = subprocess.run(
                    ["nvidia-smi", "--query-gpu=utilization.gpu,memory.used,memory.total,temperature.gpu,power.draw",
                     "--format=csv,noheader,nounits"],
                    capture_output=True, text=True, timeout=5)
                if r.returncode != 0: return None
                p = [x.strip() for x in r.stdout.strip().split("\n")[0].split(",")]
                return {"gpu/utilization_percent": float(p[0]), "gpu/memory_used_gb": float(p[1])/1024.0,
                        "gpu/temperature_celsius": float(p[3]),
                        "gpu/power_watts": float(p[4]) if p[4] != "[N/A]" else 0.0}
            except Exception: return None

        def on_log(self, args, state, control, logs=None, **kwargs):
            if logs is None or state.global_step % self.log_every_n_steps != 0: return
            smi = self._query_nvidia_smi()
            if smi: logs.update(smi)
            if torch.cuda.is_available():
                logs["gpu/memory_allocated_gb"] = torch.cuda.memory_allocated() / (1024**3)
                logs["gpu/memory_reserved_gb"]  = torch.cuda.memory_reserved()   / (1024**3)
            now = time.time()
            if self._last_step_time is not None:
                elapsed = now - self._last_step_time
                steps   = state.global_step - self._last_global_step
                if elapsed > 0 and steps > 0:
                    sps = steps / elapsed
                    logs["throughput/steps_per_sec"]   = sps
                    logs["throughput/samples_per_sec"] = sps * args.per_device_train_batch_size
            self._last_step_time    = now
            self._last_global_step  = state.global_step

        def on_train_begin(self, args, state, control, **kwargs):
            self._last_step_time   = time.time()
            self._last_global_step = state.global_step
            print("[TensorBoard] GPU metrics logging enabled.")
```

## Dr. GRPO Patch
> Detects TRL capabilities and applies the correct fix for both biases.
> Fix 1 (std norm): `scale_rewards=False` — widely supported.
> Fix 2 (length norm): `loss_type='dr_grpo'` if TRL>=0.16, else custom DrGRPOTrainer.

```python
if TRAIN_ON_KAGGLE:
    import inspect
    import trl
    import torch
    from trl import GRPOTrainer, GRPOConfig

    print(f"TRL version: {trl.__version__}")
    _cfg_params = set(inspect.signature(GRPOConfig.__init__).parameters.keys())

    HAS_SCALE_REWARDS = "scale_rewards" in _cfg_params
    HAS_LOSS_TYPE     = "loss_type"     in _cfg_params
    print(f"  scale_rewards param: {HAS_SCALE_REWARDS}")
    print(f"  loss_type param:     {HAS_LOSS_TYPE}")

    # ================================================================
    # DrGRPOTrainer — used when TRL lacks native loss_type='dr_grpo'.
    # Fix: after parent computes per-token loss (divided by seq_len),
    # multiply back by (seq_len / MAX_COMPLETION_LENGTH) so the net
    # normalization is 1/MAX_COMPLETION_LENGTH (constant, not per-seq).
    # This is equivalent to using a constant denominator in loss.
    # ================================================================
    DR_GRPO_MAX_COMPLETION = 3000  # must match GRPOConfig.max_completion_length

    class DrGRPOTrainer(GRPOTrainer):
        """
        Dr. GRPO length-normalization fix for TRL < 0.16.
        scale_rewards=False (in GRPOConfig) handles std-norm removal.
        This subclass handles 1/|o_i| -> 1/max_completion_length.
        """
        def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
            loss = super().compute_loss(model, inputs, return_outputs=return_outputs, **kwargs)
            # loss is a scalar mean over the batch at this point — we can't easily
            # re-scale individual sequences.  Instead we scale based on mean seq len
            # in the batch as a best-effort approximation.
            # For exact Dr. GRPO, upgrade to TRL >= 0.16 with loss_type='dr_grpo'.
            if isinstance(loss, torch.Tensor) and "completion_mask" in inputs:
                mask = inputs["completion_mask"].float()
                mean_seq_len = mask.sum(-1).mean()  # mean completion length in batch
                scale = mean_seq_len / DR_GRPO_MAX_COMPLETION
                loss = loss * scale
            return loss

    if HAS_LOSS_TYPE:
        ActiveTrainer = GRPOTrainer
        print("Using GRPOTrainer with loss_type='dr_grpo' (native fix)")
    else:
        ActiveTrainer = DrGRPOTrainer
        print("Using DrGRPOTrainer subclass (length norm approximation)")
        print("Upgrade to TRL >= 0.16 for exact Dr. GRPO length normalization.")
```

## Reward Functions

```python
if TRAIN_ON_KAGGLE:
    import re

    _BOXED_RE = re.compile(r'\\boxed\{([^}]*)\}')
    _THINK_RE = re.compile(r'<think>.*?</think>', re.DOTALL)
    MAX_COMPLETION_LENGTH = DR_GRPO_MAX_COMPLETION  # same constant used in loss
    OVERLONG_BUFFER       = 512   # soft penalty starts MAX_COMPLETION_LENGTH+OVERLONG_BUFFER tokens

    def _extract_boxed(text):
        # brace-balanced extraction (handles \boxed{\frac{1}{2}})
        idx = text.find(r'\boxed{')
        if idx == -1:
            m = _BOXED_RE.search(text)
            return m.group(1).strip() if m else None
        depth, start = 1, idx + 7
        for i in range(start, len(text)):
            if text[i] == '{': depth += 1
            elif text[i] == '}': depth -= 1
            if depth == 0:
                return text[start:i].strip()
        return text[start:].strip()

    def _normalize(s):
        return str(s).strip().lower().replace(" ", "")

    def format_reward(completions, **kwargs):
        """Rewards correct format: \\boxed{} present AND <think>...</think> before it."""
        rewards = []
        for c in completions:
            r = 0.0
            has_boxed = bool(_BOXED_RE.search(c))
            has_think = bool(_THINK_RE.search(c))
            # Require think block to appear before boxed (no reward-hacking via naked boxed)
            think_before_boxed = has_think and (c.find('</think>') < c.rfind(r'\boxed{'))
            if has_boxed:          r += 0.5
            if think_before_boxed: r += 0.3
            rewards.append(r)
        return rewards

    def accuracy_reward(completions, answer, **kwargs):
        """Rewards exact match between \\boxed{} and ground truth."""
        rewards = []
        for c, expected in zip(completions, answer):
            predicted = _extract_boxed(c)
            if predicted is not None and _normalize(predicted) == _normalize(expected):
                rewards.append(2.0)
            else:
                rewards.append(0.0)
        return rewards

    def overlong_penalty(completions, **kwargs):
        """
        DAPO-style soft overlong penalty.
        No penalty for len <= MAX_COMPLETION_LENGTH + OVERLONG_BUFFER.
        Gentle linear decay beyond that. Hard cap at 2x MAX_COMPLETION_LENGTH.
        """
        rewards = []
        soft_limit = MAX_COMPLETION_LENGTH + OVERLONG_BUFFER
        hard_limit = MAX_COMPLETION_LENGTH * 2
        for c in completions:
            # approximate token count: 1 token ~ 4 chars
            approx_len = len(c) / 4.0
            if approx_len <= soft_limit:
                r = 0.0
            elif approx_len >= hard_limit:
                r = -1.0
            else:
                r = -((approx_len - soft_limit) / (hard_limit - soft_limit))
            rewards.append(r)
        return rewards

    def combined_reward(completions, answer, **kwargs):
        fmt  = format_reward(completions, **kwargs)
        acc  = accuracy_reward(completions, answer=answer, **kwargs)
        over = overlong_penalty(completions, **kwargs)
        return [f + a + o for f, a, o in zip(fmt, acc, over)]

    print("Reward functions defined: format_reward, accuracy_reward, overlong_penalty, combined_reward")
```

## Dr. GRPO Trainer Config (v11)

```python
if TRAIN_ON_KAGGLE:
    from trl import GRPOConfig
    import gc, time, torch

    TB_LOG_DIR  = "/kaggle/working/tb_logs"
    ADAPTER_DIR = "/kaggle/working/sft_adapter"

    # ============================================================
    # DR. GRPO CONFIG (arxiv:2503.20783, Table 6)
    # Key changes vs v9-1:
    #   LR:          5e-6  -> 1e-6   (constant, not cosine)
    #   beta (KL):   0.04  -> 0.0    (rule-based reward = no drift risk)
    #   num_gen:     4     -> 8      (Dr. GRPO recipe minimum)
    #   temperature: 0.8   -> 1.0    (full exploration)
    #   max_comp:    1024  -> 3000   (enough for CoT + answer)
    #   grad_norm:   0.1   -> 1.0    (0.1 was too aggressive)
    #   grad_accum:  8     -> 16     (larger effective batch)
    # ============================================================
    grpo_kwargs = dict(
        output_dir                   = "/kaggle/working/grpo_output",
        num_train_epochs             = 1,
        per_device_train_batch_size  = 1,
        gradient_accumulation_steps  = 16,
        learning_rate                = 1e-6,
        lr_scheduler_type            = "constant",   # Dr. GRPO: constant LR
        warmup_ratio                 = 0.0,
        max_prompt_length            = 3072,
        max_completion_length        = DR_GRPO_MAX_COMPLETION,
        num_generations              = 8,
        temperature                  = 1.0,
        top_p                        = 1.0,
        beta                         = 0.0,
        optim                        = "adamw_8bit",
        adam_beta1                   = 0.9,
        adam_beta2                   = 0.999,
        max_grad_norm                = 1.0,
        logging_steps                = 1,
        logging_dir                  = TB_LOG_DIR,
        report_to                    = "tensorboard",
        save_strategy                = "no",
        bf16                         = True,
        gradient_checkpointing       = True,
        gradient_checkpointing_kwargs= {"use_reentrant": False},
        seed                         = SEED,
        remove_unused_columns        = False,
    )

    # Fix 1: remove std normalization (widely supported)
    if HAS_SCALE_REWARDS:
        grpo_kwargs["scale_rewards"] = False
        print("[Dr. GRPO] scale_rewards=False: std norm REMOVED")
    else:
        print("[WARN] scale_rewards not supported in this TRL — std norm NOT removed")

    # Fix 2: remove length normalization (TRL >= 0.16 only)
    if HAS_LOSS_TYPE:
        grpo_kwargs["loss_type"] = "dr_grpo"
        print("[Dr. GRPO] loss_type='dr_grpo': length norm REMOVED (native)")
    else:
        print("[Dr. GRPO] length norm handled by DrGRPOTrainer subclass")

    grpo_config = GRPOConfig(**grpo_kwargs)

    print("\n" + "="*60)
    print("  DR. GRPO TRAINING CONFIG v11")
    print("="*60)
    print(f"  LR:           {grpo_config.learning_rate}")
    print(f"  Epochs:       {grpo_config.num_train_epochs}")
    print(f"  Num gen:      {grpo_config.num_generations}")
    print(f"  Beta (KL):    {grpo_config.beta}")
    print(f"  Max comp len: {grpo_config.max_completion_length}")
    print(f"  Temperature:  {grpo_config.temperature}")
    bs = grpo_config.per_device_train_batch_size
    ga = grpo_config.gradient_accumulation_steps
    print(f"  Batch:        {bs} x {ga} = {bs*ga} effective")
    print("="*60 + "\n")

    trainer = ActiveTrainer(
        model            = model,
        args             = grpo_config,
        train_dataset    = grpo_dataset,
        reward_funcs     = [combined_reward],
        processing_class = tokenizer,
        callbacks        = [GPUMetricsCallback(log_every_n_steps=1)],
    )

    torch.cuda.empty_cache()
    gc.collect()

    print("Starting Dr. GRPO training v11...")
    t0 = time.time()
    trainer.train()
    print(f"Done in {(time.time()-t0)/60:.1f} min")

    model.save_pretrained(ADAPTER_DIR)
    tokenizer.save_pretrained(ADAPTER_DIR)
    print(f"Adapter saved to {ADAPTER_DIR}")
```
