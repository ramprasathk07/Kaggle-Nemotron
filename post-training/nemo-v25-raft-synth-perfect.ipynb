# Nemotron-3-Nano **v25** — RAFT on PERFECT synthetic data (single RTX Pro 6000, from 85%)

RAFT / Reinforce-Rejection warm-started from the **0.85 SFT adapter**, on a
**zero-label-noise** problem pool built by the `syn_datagen` deterministic solvers.

## Why this fixes the 0.67 regression
v23/v24 trained on **mixed teacher CoT** (andy + Sonnet + Tong) → inconsistent
reasoning → 0.85 → **0.67**. This notebook uses **no teacher traces**: problems are
generated synthetically and **independently verified by the syn_datagen solver**
(`compare_answer` = the official metric), so every (problem, answer, CoT) is provably
correct. The solver output *is* the ground truth — the same deterministic rules that
produced the competition's own answers.

## Data engine
- **Synthetic-perfect** (`numeral`, `gravity`, `unit_conversion`): forward-generate
  examples consistent with one rule → run the matching reasoner → keep only what it
  solves. Local yield: numeral 200/200, gravity 196/200, unit 199/200.
- **train.csv** for the hard categories (`bit_manipulation`, `cipher`, `equation`):
  perfect competition answers, exact distribution. (cipher/bit/equation solvers are
  constraint-heavy → cheaper to take real problems than to generate them.)
- Every rollout scored by the verifier; **perfect reasoner CoT** is kept as a *rescue*
  distill target for problems the policy fails on all rollouts.

## RAFT loop (per round)
1. Sample `N_ROLLOUTS` per prompt (per-prompt gen — Mamba-2 NaNs on left-pad).
2. Verify each vs ground truth.
3. Keep correct rollouts from **mixed** groups (drop all-correct = too easy,
   all-wrong → **rescue** with the solver's perfect CoT).
4. SFT the adapter on the kept corpus (gentle LR, refine not smash).

## Single-GPU compute budget
Synthetic CoT is short → small knobs fit fast. Defaults: `N_PROMPTS=600`,
`N_ROLLOUTS=4`, `GEN_MAX_NEW=1024`, `RAFT_LR=2e-5`, 1 round ≈ ~1h. Scale via the
config cell. `SMOKE_TEST=1` first (8 prompts) to confirm rollouts hit `\boxed{}`.

> RL is the path past the SFT ceiling (LoRA Without Regret: RL matches full-FT even at
> rank 1 → rank-32 is plenty). Start here from 0.85, eval-gate every round.

## Verifiable Reward (deterministic, per-type)

Ported verbatim from the GRPO notebook: classify puzzle type, brace-balanced
`\boxed{}` extraction, type-aware comparison (numeric tolerance for gravity/units,
multi-base integer match for bit/base, case-insensitive for cipher).


```python
import re

_PUZZLE_PATTERNS = [
    ("Number Base Conversion", re.compile(r"numeral system|base[- ]?\d|number.*convert|radix|secret number", re.IGNORECASE)),
    ("Gravitational Constant", re.compile(r"gravit|gravity|falling|free.?fall|acceleration due to", re.IGNORECASE)),
    ("Equation Transformation", re.compile(r"transformation rule|equation.*transform|secret.*rule.*equation|rule.*applied.*equation", re.IGNORECASE)),
    ("Text Encryption",         re.compile(r"encrypt|cipher|secret.*code.*letter|coded.*message|secret.*text", re.IGNORECASE)),
    ("Bit Manipulation",        re.compile(r"bit.?manipul|binary|8.?bit|bitwise|bit.*transform", re.IGNORECASE)),
    ("Unit Conversion",         re.compile(r"unit.?conver|measurement|becomes.*\d|secret.*conver.*measur", re.IGNORECASE)),
]

def classify_puzzle(prompt: str) -> str:
    for label, pat in _PUZZLE_PATTERNS:
        if pat.search(prompt or ""):
            return label
    return "Unknown"

_BOXED_RE = re.compile(r"\\boxed\{([^}]*)\}")
_THINK_RE = re.compile(r"<think>.*?</think>", re.DOTALL)

def extract_boxed(text):
    idx = text.find("\\boxed{")
    if idx == -1:
        m = _BOXED_RE.search(text)
        return m.group(1).strip() if m else None
    depth, start = 1, idx + 7
    for i in range(start, len(text)):
        if text[i] == "{":   depth += 1
        elif text[i] == "}": depth -= 1
        if depth == 0:       return text[start:i].strip()
    return text[start:].strip()

def _norm(s):
    return str(s).strip().lower().replace(" ", "").replace(",", "")

def _numeric_match(predicted, expected, rel_tol=1e-2, abs_tol=1e-4):
    try:
        pf = float(str(predicted).replace(",", "").strip())
        ef = float(str(expected).replace(",", "").strip())
    except (ValueError, TypeError):
        return False
    return abs(pf - ef) <= max(rel_tol * max(1.0, abs(ef)), abs_tol)

def _integer_match(predicted, expected, bases=(2,8,10,16)):
    p_str, e_str = _norm(predicted), _norm(expected)
    def parse(s, base):
        try:
            if base == 16 and s.startswith("0x"): s = s[2:]
            elif base == 2 and s.startswith("0b"): s = s[2:]
            return int(s, base)
        except ValueError:
            return None
    for b1 in bases:
        pv = parse(p_str, b1)
        if pv is None: continue
        for b2 in bases:
            ev = parse(e_str, b2)
            if ev is not None and pv == ev:
                return True
    return False

def verify_answer(predicted, expected, puzzle_type):
    if predicted is None:
        return False
    pn, en = _norm(predicted), _norm(expected)
    if pn == en:
        return True
    if puzzle_type in ("Gravitational Constant", "Unit Conversion"):
        return _numeric_match(predicted, expected, rel_tol=1e-2, abs_tol=1e-4)
    if puzzle_type in ("Bit Manipulation", "Number Base Conversion"):
        return _integer_match(predicted, expected)
    if puzzle_type == "Text Encryption":
        return pn.replace("'", "") == en.replace("'", "")
    if puzzle_type == "Equation Transformation":
        return _numeric_match(predicted, expected) or pn == en
    return _numeric_match(predicted, expected)

def has_think_before_boxed(text):
    has_box = bool(_BOXED_RE.search(text)) or ("\\boxed{" in text)
    has_think = bool(_THINK_RE.search(text))
    return has_think and has_box and (text.find("</think>") < text.rfind("\\boxed{"))

# smoke
assert verify_answer("42", "42", "Unknown")
assert verify_answer("9.81", "9.80", "Gravitational Constant")
assert verify_answer("0xFF", "255", "Number Base Conversion")
assert verify_answer("HELLO", "hello", "Text Encryption")
print("Verifier smoke tests pass.")
```

```python
from __future__ import annotations
# ── v25: syn_datagen solvers VENDORED INLINE (no dataset upload needed) ──
# Exact copies of syn_datagen/{store_types,numeral,gravity,unit_conversion,reasoning}.py
# for the categories v25 generates synthetically (numeral/gravity/unit). The hard
# categories (bit/cipher/equation) come from train.csv, so their solvers aren't needed here.
import re, math


class Example:
    def __init__(self, input_value: str, output_value: str):
        self.input_value = input_value
        self.output_value = output_value


class Problem:
    def __init__(self, id, category, examples, question, answer, prompt=""):
        self.id = id; self.category = category; self.examples = examples
        self.question = question; self.answer = answer; self.prompt = prompt


def _fmt_int_with_dp(value: int, dp: int) -> str:
    if dp == 0:
        return str(value)
    s = str(value).zfill(dp + 1)
    s = s[: len(s) - dp] + "." + s[len(s) - dp :]
    s = s.lstrip("0") or "0"
    if s.startswith("."):
        s = "0" + s
    return s


def truncate_3dp(s: str) -> str:
    if "." not in s:
        return s
    integer, frac = s.split(".")
    if len(frac) <= 3:
        return s
    return integer + "." + frac[:3]


def _dp_count(s: str) -> int:
    return 0 if "." not in s else len(s.split(".")[1])


def pad_dp(s: str, n: int) -> str:
    if "." not in s:
        s = s + "."
    integer, frac = s.split(".")
    return integer + "." + frac.ljust(n, "0")


def cast_dp_pair(a: str, b: str):
    da, db = _dp_count(a), _dp_count(b)
    target = max(da, db)
    return pad_dp(a, target), pad_dp(b, target), target, target


def long_multiplication_lines(a_str: str, b_str: str):
    a_dp = len(a_str.split(".")[1]) if "." in a_str else 0
    b_dp = len(b_str.split(".")[1]) if "." in b_str else 0
    total_dp = a_dp + b_dp
    a_int = int(a_str.replace(".", ""))
    b_int = int(b_str.replace(".", ""))
    lines = []
    b_digits_str = str(abs(b_int))
    b_num_digits = len(b_digits_str)
    components = []
    for i in range(b_num_digits - 1, -1, -1):
        d = int(b_digits_str[i])
        if d == 0:
            continue
        comp_scaled = d * (10 ** (b_num_digits - 1 - i))
        comp_display = _fmt_int_with_dp(comp_scaled, b_dp)
        if b_dp > 0:
            comp_display = pad_dp(comp_display, b_dp)
        product_int = a_int * comp_scaled
        product_display = _fmt_int_with_dp(product_int, total_dp)
        if total_dp > 0:
            product_display = pad_dp(product_display, total_dp)
        components.append((comp_display, product_int, product_display))
    for comp_display, _, product_display in components:
        lines.append(f"{a_str} * {comp_display} = {product_display}")
    if len(components) >= 2:
        running = components[0][1]
        for i in range(1, len(components)):
            running_display = _fmt_int_with_dp(running, total_dp)
            if total_dp > 0:
                running_display = pad_dp(running_display, total_dp)
            running += components[i][1]
            sum_display = _fmt_int_with_dp(running, total_dp)
            if total_dp > 0:
                sum_display = pad_dp(sum_display, total_dp)
            lines.append(f"{running_display} + {components[i][2]} = {sum_display}")
    total = a_int * b_int
    result_str = _fmt_int_with_dp(total, total_dp)
    return lines, result_str


def long_division_lines(numerator_str: str, denominator_str: str, max_decimal_digits: int = 3):
    n_dp = len(numerator_str.split(".")[1]) if "." in numerator_str else 0
    d_dp = len(denominator_str.split(".")[1]) if "." in denominator_str else 0
    max_dp = max(n_dp, d_dp)
    num = int(round(float(numerator_str) * 10**max_dp))
    den = int(round(float(denominator_str) * 10**max_dp))
    lines = []
    acc = 0
    decimal_digits = 0

    def fmt_acc():
        if decimal_digits == 0:
            return str(acc)
        s = str(acc).zfill(decimal_digits + 1)
        return s[:-decimal_digits] + "." + s[-decimal_digits:]

    def fmt_scale():
        if decimal_digits == 0:
            return "1"
        return "0." + "0" * (decimal_digits - 1) + "1"

    def fmt_line(n):
        return f"= {fmt_acc()} + {fmt_scale()} * {n} / {den}"

    lines.append(fmt_line(num))
    while decimal_digits <= max_decimal_digits:
        if num >= den:
            num -= den; acc += 1
            lines.append(fmt_line(num))
        else:
            decimal_digits += 1
            if decimal_digits > max_decimal_digits:
                break
            num *= 10; acc *= 10
            lines.append(fmt_line(num))
    if decimal_digits > max_decimal_digits:
        decimal_digits = max_decimal_digits
    return lines, fmt_acc()


# ── numeral ──
ROMAN_VALUES = [(1000, "M"), (900, "CM"), (500, "D"), (400, "CD"), (100, "C"),
                (90, "XC"), (50, "L"), (40, "XL"), (10, "X"), (9, "IX"),
                (5, "V"), (4, "IV"), (1, "I")]


def _to_roman(n: int) -> str:
    parts = []; remaining = n
    for val, sym in ROMAN_VALUES:
        while remaining >= val:
            parts.append(sym); remaining -= val
    return "".join(parts)


def reasoning_numeral(problem) -> str:
    lines = ["This is an Arabic to Roman numeral conversion.", ""]
    n = int(problem.question)
    computed = _to_roman(n)
    lines.append(f"Converting {n}:")
    remaining = n; parts = []
    for val, sym in ROMAN_VALUES:
        while remaining >= val:
            lines.append(f"  {remaining} >= {val} -> {sym}, remainder {remaining - val}")
            parts.append(sym); remaining -= val
    lines.append("")
    lines.append(f"Result: {' '.join(parts)} -> {{{computed}}}")
    lines.append("")
    lines.append("I will now return the answer in \\boxed{}")
    lines.append(f"The answer in \\boxed{{–}} is \\boxed{{{computed}}}")
    return "\n".join(lines)


# ── gravity: d = k * t^2 ──
def reasoning_gravity(problem):
    lines = ["We need to determine the falling distance using d = k*t^2. Let me find k from the examples.",
             "I will put my final answer inside \\boxed{}.", ""]
    k_strs = []
    for ex in problem.examples:
        t = float(ex.input_value)
        if t > 0:
            t_squared = round(t * t, 4)
            t_sq_full = str(t_squared)
            t_sq_str = truncate_3dp(t_sq_full)
            d_str = truncate_3dp(ex.output_value)
            lines.append(f"t = {ex.input_value}s, d = {ex.output_value}m:")
            lines.append(f"t^2 = {ex.input_value} * {ex.input_value}:")
            sq_lines, sq_result = long_multiplication_lines(ex.input_value, ex.input_value)
            lines.extend(sq_lines)
            if sq_result != t_sq_full:
                lines.append(f"= {t_sq_full}")
            d_cast, tsq_cast, _, _ = cast_dp_pair(d_str, t_sq_str)
            lines.append(f"k = {ex.output_value} / {ex.input_value}^2 = {d_str} / {t_sq_full} = {d_cast} / {tsq_cast}")
            div_lines, k_str = long_division_lines(d_cast, tsq_cast)
            lines.extend(div_lines); lines.append(f"= {k_str}")
            k_strs.append(k_str); lines.append("")
    if not k_strs:
        return None
    k_values = [float(s) for s in k_strs]
    lines.append(f"k values: {', '.join(k_strs)}")
    paired = sorted(zip(k_values, k_strs))
    lines.append(f"k values (sorted): {', '.join(s for _, s in paired)}")
    if len(paired) % 2 == 0 and len(paired) >= 2:
        _, k_fit_str = paired[len(paired) // 2 - 1]
    else:
        _, k_fit_str = paired[len(paired) // 2]
    lines.append(f"The median k is {k_fit_str}.")
    lines.append("")
    lines.append(f"For t = {problem.question}:")
    lines.append(f"t^2 = {problem.question} * {problem.question}:")
    sq_lines, t_sq_str = long_multiplication_lines(problem.question, problem.question)
    lines.extend(sq_lines); lines.append(f"= {t_sq_str}"); lines.append("")
    k_display = k_fit_str.rstrip("0").rstrip(".")
    lines.append(f"d = {k_display} * {t_sq_str}:")
    mult_lines, mult_result = long_multiplication_lines(k_display, t_sq_str)
    lines.extend(mult_lines)
    dot = mult_result.index("."); boxed_answer = mult_result[: dot + 4]
    lines.append(f"= {boxed_answer}"); lines.append("")
    lines.append("I will now return the answer in \\boxed{}")
    lines.append(f"The answer in \\boxed{{–}} is \\boxed{{{boxed_answer}}}")
    return "\n".join(lines)


# ── unit conversion: output = factor * input ──
def reasoning_unit_conversion(problem):
    lines = ["We need to find a conversion rule that maps the inputs to outputs. Let me check if it's a linear factor.",
             "I will put my final answer inside \\boxed{}.", ""]
    factor_strs = []
    for ex in problem.examples:
        inp = float(ex.input_value)
        if inp != 0:
            out_str = truncate_3dp(ex.output_value)
            inp_str = truncate_3dp(ex.input_value)
            lines.append(f"{ex.input_value} -> {ex.output_value}")
            inp_cast, out_cast, inp_dp, out_dp = cast_dp_pair(inp_str, out_str)
            lines.append(f"Casting input to {inp_dp} decimal places, output to {out_dp} decimal places: {inp_cast} -> {out_cast}")
            lines.append(f"factor = {out_cast} / {inp_cast}")
            div_lines, factor_str = long_division_lines(out_cast, inp_cast)
            lines.extend(div_lines); lines.append(f"= {factor_str}")
            factor_strs.append(factor_str); lines.append("")
    if not factor_strs:
        return None
    factors = [float(s) for s in factor_strs]
    lines.append(f"factor values: {', '.join(factor_strs)}")
    paired = sorted(zip(factors, factor_strs))
    lines.append(f"factor values (sorted): {', '.join(s for _, s in paired)}")
    if len(paired) % 2 == 0 and len(paired) >= 2:
        _, med_factor_str = paired[len(paired) // 2 - 1]
    else:
        _, med_factor_str = paired[len(paired) // 2]
    lines.append(f"The median factor is {med_factor_str}.")
    med_display = med_factor_str.rstrip("0").rstrip(".")
    lines.append("")
    lines.append(f"Converting {problem.question}:")
    lines.append(f"{problem.question} * {med_display}:")
    mult_lines, mult_result = long_multiplication_lines(problem.question, med_display)
    lines.extend(mult_lines)
    dot = mult_result.index("."); boxed_answer = mult_result[: dot + 4]
    lines.append(f"= {boxed_answer}"); lines.append("")
    lines.append("I will now return the answer in \\boxed{}")
    lines.append(f"The answer in \\boxed{{–}} is \\boxed{{{boxed_answer}}}")
    return "\n".join(lines)


# ── verifier (official metric) ──
def _reasoner_extract(reasoning_text: str) -> str:
    matches = re.findall(r"\\boxed\{([^}]*)(?:\}|$)", reasoning_text)
    if matches:
        non_empty = [m.strip() for m in matches if m.strip()]
        return non_empty[-1] if non_empty else matches[-1].strip()
    return ""


def compare_answer(stored_answer: str, predicted: str) -> bool:
    stored_answer = str(stored_answer).strip(); predicted = str(predicted).strip()
    if re.fullmatch(r"[01]+", stored_answer):
        return predicted.lower() == stored_answer.lower()
    try:
        return math.isclose(float(stored_answer), float(predicted), rel_tol=1e-2, abs_tol=1e-5)
    except Exception:
        return predicted.lower() == stored_answer.lower()


# smoke
assert _to_roman(38) == "XXXVIII"
assert compare_answer("24.64", "24.6401")
print("reasoners vendored inline OK (numeral/gravity/unit + verifier)")
```

```python
# ── v25 DATA ENGINE: perfect synthetic (solver-verified) + train.csv hard cats ──
import random, re
import pandas as pd
from collections import Counter

random.seed(SEED)
_BX = re.compile(r"\\boxed\{([^{}]*)\}")

def _boxed(t):
    if not t:
        return None
    m = _BX.findall(t)
    return m[-1].strip() if m else None

def _strip_box(t):
    """Remove every \\boxed{...} (brace-balanced) from a reasoning body."""
    tok, out, i = "\\boxed{", [], 0
    while i < len(t):
        j = t.find(tok, i)
        if j == -1:
            out.append(t[i:]); break
        out.append(t[i:j]); k = j + len(tok); d = 1
        while k < len(t) and d > 0:
            if t[k] == "{": d += 1
            elif t[k] == "}": d -= 1
            k += 1
        i = k
    return "".join(out)

def _to_target(reasoner_text, ans):
    """Solver CoT -> canonical SFT target <think>..</think>\\boxed{ans}."""
    t = reasoner_text
    idx = t.find("I will now return the answer")
    if idx != -1:
        t = t[:idx]
    t = _strip_box(t).strip()
    t = re.sub(r"\n{3,}", "\n\n", t)
    if not t:
        t = "Work through the problem step by step."
    return f"<think>\n{t}\n</think>\n\\boxed{{{ans}}}"

# ---- forward generators: build examples consistent with ONE rule, then let the
#      matching solver compute the ground-truth answer + perfect CoT ----
def gen_numeral():
    n = random.randint(1, 3999)
    seen, exs = {n}, []
    while len(exs) < 4:
        m = random.randint(1, 3999)
        if m in seen:
            continue
        seen.add(m); exs.append(Example(str(m), ""))
    r = reasoning_numeral(Problem("syn", "numeral", exs, str(n), ""))
    ans = _boxed(r)
    if ans is None:
        return None
    ex_lines = "\n".join(f"{e.input_value} -> {_to_roman(int(e.input_value))}" for e in exs)
    prompt = ("In Alice's Wonderland, numbers are secretly converted into a different "
              "numeral system. Some examples are given below:\n" + ex_lines +
              f"\nNow, write the number {n} in the Wonderland numeral system.")
    return prompt, ans, r

def gen_gravity():
    k = round(random.uniform(4.0, 12.0), 2)           # k = 0.5*g
    ts = set()
    while len(ts) < 5:
        ts.add(round(random.uniform(1.0, 5.0), 2))
    exs = [Example(f"{t:.2f}", f"{round(k*t*t,2):.2f}") for t in sorted(ts)]
    qt = round(random.uniform(1.0, 5.0), 2)
    r = reasoning_gravity(Problem("syn", "gravity", exs, f"{qt:.2f}", ""))
    ans = _boxed(r)
    if ans is None:
        return None
    ex_lines = "\n".join(f"For t = {e.input_value}s, distance = {e.output_value} m" for e in exs)
    prompt = ("In Alice's Wonderland, the gravitational constant has been secretly changed. "
              "Here are some example observations:\n" + ex_lines +
              f"\nNow, determine the falling distance for t = {qt:.2f}s given d = 0.5*g*t^2.")
    return prompt, ans, r

def gen_unit():
    f = round(random.uniform(0.3, 3.0), 3)
    ins = set()
    while len(ins) < 5:
        ins.add(round(random.uniform(5.0, 40.0), 2))
    exs = [Example(f"{i:.2f}", f"{round(f*i,2):.2f}") for i in sorted(ins)]
    qi = round(random.uniform(5.0, 40.0), 2)
    r = reasoning_unit_conversion(Problem("syn", "unit_conversion", exs, f"{qi:.2f}", ""))
    ans = _boxed(r)
    if ans is None:
        return None
    ex_lines = "\n".join(f"{e.input_value} m becomes {e.output_value}" for e in exs)
    prompt = ("In Alice's Wonderland, a secret unit conversion is applied to measurements. "
              "For example:\n" + ex_lines + f"\nNow, convert the following measurement: {qi:.2f} m")
    return prompt, ans, r

_GEN = {"numeral": gen_numeral, "gravity": gen_gravity, "unit_conversion": gen_unit}

prompts = []
PERFECT_COT = {}     # raw_prompt -> canonical perfect target (rescue distill)

for cat in SYN_CATEGORIES:
    g = _GEN[cat]; got = tries = 0
    while got < SYN_PER_CAT and tries < SYN_PER_CAT * 8:
        tries += 1
        res = g()
        if not res:
            continue
        pr, ans, rtext = res
        if pr in PERFECT_COT:
            continue
        # double-check the solver answer round-trips through the official metric
        if not compare_answer(ans, _boxed(rtext) or ""):
            continue
        prompts.append({"raw_prompt": pr, "answer": ans})
        PERFECT_COT[pr] = _to_target(rtext, ans)
        got += 1
    print(f"[synth] {cat}: {got}/{SYN_PER_CAT} verified-perfect")

if USE_TRAINCSV_HARD:
    def _cl(p):
        p = p.lower()
        if "bit manipulation" in p or "8-bit" in p: return "bit_manipulation"
        if "encrypt" in p or "decrypt" in p or "cipher" in p: return "cipher"
        if "transformation rule" in p or "equation" in p: return "equation"
        return "other"
    df = pd.read_csv(RAFT_DATA_PATH).dropna(subset=["prompt", "answer"])
    df = df.sample(frac=1, random_state=SEED).reset_index(drop=True)
    df["c"] = df["prompt"].map(_cl)
    for cat in HARD_CATEGORIES:
        sub = df[df["c"] == cat].head(HARD_PER_CAT)
        for _, r in sub.iterrows():
            prompts.append({"raw_prompt": str(r["prompt"]), "answer": str(r["answer"])})
        print(f"[traincsv] {cat}: {len(sub)}")

random.shuffle(prompts)
print(f"\nTOTAL pool: {len(prompts)} prompts | perfect-CoT available for {len(PERFECT_COT)}")
print("type dist:", dict(Counter(classify_puzzle(p["raw_prompt"]) for p in prompts)))
print("--- sample synthetic prompt ---\n", prompts[0]["raw_prompt"][:280])
```

## Sample N rollouts per prompt, verify, cache to disk

Generation memory is toggled safely (gradient checkpointing OFF + KV cache ON for
gen, restored after). Rollouts + correctness are written to `ROLLOUT_CACHE` so a
later SFT crash never costs the expensive sampling. If the cache already exists it
is reused (delete it to re-sample).


```python
import json, time, gc, torch
from transformers import LogitsProcessor, LogitsProcessorList

def build_gen_prompt(raw_prompt):
    msgs = [{"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user",   "content": raw_prompt + PROMPT_SUFFIX}]
    try:
        return tokenizer.apply_chat_template(msgs, tokenize=False,
                                             add_generation_prompt=True, enable_thinking=True)
    except TypeError:
        return tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)

class SanitizeLogits(LogitsProcessor):
    """Safety net: replace non-finite logits so torch.multinomial never asserts.
    A clean run should NEVER trigger this; we count hits to surface silent NaN
    instead of letting `probability tensor contains inf/nan` crash the loop."""
    def __init__(self):
        self.hits = 0
    def __call__(self, input_ids, scores):
        if not torch.isfinite(scores).all():
            self.hits += 1
            scores = torch.nan_to_num(scores, nan=-1e4, posinf=1e4, neginf=-1e4)
        return scores

def _set_gen_mode(on):
    try:
        if on: model.gradient_checkpointing_disable()
        else:  model.gradient_checkpointing_enable()
    except Exception as e:
        print("grad-ckpt toggle warn:", e)
    try:
        model.config.use_cache = on
    except Exception:
        try: model.base_model.config.use_cache = on
        except Exception: pass

if os.path.exists(ROLLOUT_CACHE) and not SMOKE_TEST:
    print(f"Rollout cache exists -> reusing {ROLLOUT_CACHE} (delete to re-sample).")
else:
    if os.path.exists(ROLLOUT_CACHE):
        os.remove(ROLLOUT_CACHE)
    _set_gen_mode(True)
    model.eval()

    # ---- PER-PROMPT generation, NO cross-prompt padding ---------------------
    # Nemotron-3-Nano is a hybrid Mamba-2 / attention model. Mamba SSM layers
    # corrupt their state on LEFT-padded batches (conv1d + selective scan run
    # over the pad positions) -> logits become NaN -> softmax/multinomial throws
    #   "probability tensor contains either inf, nan or element < 0".
    # That is exactly the AcceleratorError this cell used to hit under
    # do_sample=True. Fix: generate ONE prompt per call and get the N rollouts
    # via num_return_sequences -- every returned sequence shares an identical
    # prompt length, so there is no padding and the Mamba scan stays finite.
    # (Cross-prompt BATCH_PROMPTS batching is intentionally NOT used here.)
    logits_guard = SanitizeLogits()
    lp = LogitsProcessorList([logits_guard])

    torch.cuda.empty_cache(); gc.collect(); torch.cuda.reset_peak_memory_stats()
    t0 = time.time()
    n_done = n_correct_rollouts = 0
    _report_every = max(1, len(prompts) // 20)

    fout = open(ROLLOUT_CACHE, "w", encoding="utf-8")
    try:
        for ex in prompts:
            text = build_gen_prompt(ex["raw_prompt"])
            enc = tokenizer(text, return_tensors="pt", truncation=True,
                            max_length=TRAIN_MAX_LEN).to(model.device)
            plen = enc["input_ids"].shape[1]     # single sequence -> no padding

            with torch.no_grad():
                out = model.generate(
                    **enc,
                    max_new_tokens=GEN_MAX_NEW,
                    do_sample=True,
                    temperature=GEN_TEMP,
                    top_p=GEN_TOP_P,
                    num_return_sequences=N_ROLLOUTS,
                    pad_token_id=tokenizer.pad_token_id,
                    logits_processor=lp,
                )
            # out: (N_ROLLOUTS, plen + gen_len). Generation starts at column plen.

            ptype = classify_puzzle(ex["raw_prompt"])
            comps, flags = [], []
            for seq in out:
                gen = tokenizer.decode(seq[plen:], skip_special_tokens=True)
                pred = extract_boxed(gen)
                ok = bool(verify_answer(pred, ex["answer"], ptype))
                comps.append(gen); flags.append(ok)
                n_correct_rollouts += int(ok)
            if n_done == 0:
                _c0 = comps[0]
                print("[rollout-dbg] rollout0 tail:", repr(_c0[-220:]))
                print("[rollout-dbg] pred=", repr(extract_boxed(_c0)), "gold=", repr(ex["answer"]),
                      "has_think=", ("</think>" in _c0), "has_boxed=", ("oxed{" in _c0), "ok=", flags[0])
            fout.write(json.dumps({
                "raw_prompt": ex["raw_prompt"], "answer": ex["answer"],
                "ptype": ptype, "completions": comps, "correct": flags,
            }) + "\n")
            n_done += 1

            del out, enc
            torch.cuda.empty_cache()
            if n_done % _report_every == 0 or n_done == len(prompts):
                el = time.time() - t0
                print(f"  [{n_done}/{len(prompts)}] {el/60:.1f} min  "
                      f"~{el/max(1,n_done):.1f}s/prompt  corr_rollouts={n_correct_rollouts}  "
                      f"NaN-guard={logits_guard.hits}  "
                      f"peakVRAM={torch.cuda.max_memory_allocated()/1e9:.1f}GB")
    finally:
        fout.close()
        _set_gen_mode(False)
        model.train()

    el = time.time() - t0
    print(f"\nSampling done: {n_done} prompts x {N_ROLLOUTS} in {el/60:.1f} min")
    print(f"EST for full set: {el/max(1,n_done):.1f}s/prompt -> "
          f"9500 prompts ~= {9500*el/max(1,n_done)/3600:.1f} h at these knobs")
    print(f"Hit rate: {n_correct_rollouts}/{n_done*N_ROLLOUTS} rollouts correct "
          f"({100*n_correct_rollouts/max(1,n_done*N_ROLLOUTS):.1f}%)")
    print(f"NaN-guard hits: {logits_guard.hits}  (MUST be 0; >0 means logits still "
          f"go non-finite -> check padding / cache / dtype before trusting rollouts)")
    print(f"PEAK VRAM (gen): {torch.cuda.max_memory_allocated()/1e9:.1f} GB")
    if n_correct_rollouts == 0:
        print("\n[WARN] 0 correct rollouts. Cold-start cannot bootstrap RAFT on this benchmark. "
              "Either (a) warm-start from an SFT adapter (set SFT_ADAPTER_DIR), or "
              "(b) raise N_ROLLOUTS to 8+ and try again, or "
              "(c) use easier prompts (filter by p_type or seed).")
```

## Filter rollouts -> SFT corpus (RAFT / Reinforce-Rej)

Per prompt, pass-rate `p = #correct / N`. `raft` keeps top-k correct completions
from any prompt with >=1 correct. `reinforce_rej` additionally drops `p==1`
(all-correct) prompts -- those carry no learning signal (Xiong 2025). Kept
completions become the assistant targets (self-distillation).


```python
import json
from datasets import Dataset as HFDataset

groups = []
with open(ROLLOUT_CACHE, encoding="utf-8") as f:
    for line in f:
        groups.append(json.loads(line))


def build_corpus(mode, keep_format, allow_any_completion=False):
    """RAFT / Reinforce-Rej filtering + v25 PERFECT-CoT rescue.

    mode='raft' keeps any group with >=1 correct; 'reinforce_rej' also drops
    all-correct (p=1) groups. On ALL-WRONG groups, if PERFECT_RESCUE and a verified
    solver CoT exists for the prompt, distill THAT (teach the perfect solution on the
    problems the policy can't yet solve).
    """
    n_all_wrong = n_all_correct = n_mixed = n_rescued = 0
    out = []
    for g in groups:
        flags = g["correct"]
        n = len(flags); n_ok = sum(flags)
        p = n_ok / max(1, n)
        if n_ok == 0:
            n_all_wrong += 1
            if PERFECT_RESCUE and g["raw_prompt"] in PERFECT_COT:
                out.append({
                    "system": SYSTEM_PROMPT,
                    "user": g["raw_prompt"] + PROMPT_SUFFIX,
                    "assistant": PERFECT_COT[g["raw_prompt"]],
                })
                n_rescued += 1
                continue
            if not allow_any_completion:
                continue
            cand = list(g["completions"])
        else:
            if p >= 1.0:
                n_all_correct += 1
                if mode == "reinforce_rej":
                    continue
            else:
                n_mixed += 1
            cand = [c for c, ok in zip(g["completions"], flags) if ok]

        if keep_format:
            fmt = [c for c in cand if has_think_before_boxed(c)]
            cand = fmt if fmt else cand
        cand = sorted(cand, key=len)[:TOP_K_KEEP]

        for comp in cand:
            out.append({
                "system": SYSTEM_PROMPT,
                "user": g["raw_prompt"] + PROMPT_SUFFIX,
                "assistant": comp.strip(),
            })
    return out, dict(all_wrong=n_all_wrong, all_correct=n_all_correct,
                     mixed=n_mixed, rescued=n_rescued)


records, stats = build_corpus(RAFT_MODE, KEEP_FORMAT_VALID)
print(f"[try1] mode={RAFT_MODE} format={KEEP_FORMAT_VALID} -> corpus={len(records)} groups={stats}")

if len(records) == 0 and KEEP_FORMAT_VALID:
    records, stats = build_corpus(RAFT_MODE, False)
    print(f"[try2] format=False -> corpus={len(records)} groups={stats}")

if len(records) == 0 and RAFT_MODE == "reinforce_rej":
    records, stats = build_corpus("raft", False)
    print(f"[try3] mode=raft -> corpus={len(records)} groups={stats}")

if len(records) == 0:
    raise RuntimeError(
        "Empty corpus after all fallbacks. With PERFECT_RESCUE on, this means the "
        "synthetic pool produced no perfect CoT AND no correct rollouts -- check the "
        "reasoners import (cell) and that warm-start loaded the 0.85 adapter.")

raw_ds = HFDataset.from_list(records)
print(f"\nFinal RAFT corpus: {len(records)} examples "
      f"(rescued-by-solver: {stats.get('rescued', 0)})")
print("--- sample target ---\n", records[0]["assistant"][:400])
```

```python
def tokenize_with_assistant_mask(example):
    full_msgs = [
        {"role": "system",    "content": example["system"]},
        {"role": "user",      "content": example["user"]},
        {"role": "assistant", "content": example["assistant"]},
    ]
    prefix_msgs = [
        {"role": "system", "content": example["system"]},
        {"role": "user",   "content": example["user"]},
    ]

    def render(msgs, add_gen):
        try:
            return tokenizer.apply_chat_template(
                msgs, tokenize=False, add_generation_prompt=add_gen, enable_thinking=True)
        except TypeError:
            return tokenizer.apply_chat_template(
                msgs, tokenize=False, add_generation_prompt=add_gen)

    full_text   = render(full_msgs, False)
    prefix_text = render(prefix_msgs, True)

    full_ids   = tokenizer(full_text, add_special_tokens=False, truncation=True,
                           max_length=TRAIN_MAX_LEN)["input_ids"]
    prefix_ids = tokenizer(prefix_text, add_special_tokens=False)["input_ids"]

    cutoff = min(len(prefix_ids), len(full_ids))
    labels = list(full_ids)
    for i in range(cutoff):
        labels[i] = -100
    return {"input_ids": full_ids, "labels": labels}


tokenized_ds = raw_ds.map(tokenize_with_assistant_mask,
                          remove_columns=raw_ds.column_names,
                          desc="Tokenize + assistant mask")

def _has_label(ex):
    return any(t != -100 for t in ex["labels"])
before = len(tokenized_ds)
tokenized_ds = tokenized_ds.filter(_has_label)
print(f"Kept {len(tokenized_ds)}/{before} rows with >=1 unmasked assistant token.")

# ---- HARD GUARD: empty after filter -> trainer would log loss 0.0 on no data --
# Rollouts that open <think> but never close </think> get fully masked by the
# prefix trick (the thinking template mangles them) and are dropped here. If that
# wipes the corpus, fail loudly instead of "training" on nothing.
if len(tokenized_ds) == 0:
    raise RuntimeError(
        f"tokenized_ds EMPTY after masking ({before} rows in, 0 kept). Every kept "
        "rollout was fully masked -> the chat template stripped assistant content "
        "(rollouts likely lack a closing </think>). Re-sample with a warm-start "
        "adapter, or set KEEP_FORMAT_VALID=True so only well-formed rollouts enter."
    )
if before - len(tokenized_ds) > 0.30 * before:
    print(f"[WARN] {before-len(tokenized_ds)} rows ({(before-len(tokenized_ds))/before*100:.0f}%) "
          f"dropped as fully-masked -> many rollouts are malformed (<think> not closed).")

# mask-check: the unmasked span must be the assistant completion (reasoning + box)
_ex0 = tokenized_ds[0]
_un = [t for t, l in zip(_ex0["input_ids"], _ex0["labels"]) if l != -100]
print(f"[mask-check] row0 total={len(_ex0['input_ids'])} unmasked={len(_un)}")
print("[mask-check] decoded unmasked:", repr(tokenizer.decode(_un)[:200]))

import numpy as np
_lens = np.array([len(x) for x in tokenized_ds["input_ids"]])
pct = lambda p: int(np.percentile(_lens, p))
print(f"Token length  min={_lens.min()}  mean={_lens.mean():.0f}  "
      f"p50={pct(50)}  p90={pct(90)}  p99={pct(99)}  max={_lens.max()}")
```

```python
import torch

class CompletionOnlyDataCollator:
    """Pads input_ids/labels/attention_mask; preserves -100 on prompt tokens."""
    def __init__(self, tokenizer, label_pad_id=-100):
        self.pad_id = tokenizer.pad_token_id
        self.label_pad_id = label_pad_id

    def __call__(self, features):
        maxlen = max(len(f["input_ids"]) for f in features)
        input_ids, labels, attn = [], [], []
        for f in features:
            ids = list(f["input_ids"]); lab = list(f["labels"])
            pad = maxlen - len(ids)
            input_ids.append(ids + [self.pad_id] * pad)
            labels.append(lab + [self.label_pad_id] * pad)
            attn.append([1] * len(ids) + [0] * pad)
        return {
            "input_ids": torch.tensor(input_ids, dtype=torch.long),
            "attention_mask": torch.tensor(attn, dtype=torch.long),
            "labels": torch.tensor(labels, dtype=torch.long),
        }

# SFT uses RIGHT padding for causal LM loss (we set LEFT earlier for generation).
tokenizer.padding_side = "right"
data_collator = CompletionOnlyDataCollator(tokenizer)
print("Collator ready. padding_side =", tokenizer.padding_side)
```

## Vanilla TRL SFT + memory-safe loss

Same fix as the other notebooks: purge Unsloth (its fused loss double-projects
lm_head on Nemotron-H) and subclass vanilla `SFTTrainer` with fused
`cross_entropy` (no fp32 full-vocab log_softmax spike). RAFT trains with plain
mean NLL on the kept correct completions.


```python
import os, sys
os.environ["TORCHDYNAMO_DISABLE"]   = "1"
os.environ["TORCH_COMPILE_DISABLE"] = "1"
os.environ["PYTORCH_ALLOC_CONF"]    = "expandable_segments:True"

import torch, torch._dynamo
import torch.nn.functional as F
torch._dynamo.config.disable = True
torch._dynamo.reset()

for _m in list(sys.modules):
    if _m == "trl" or _m.startswith("trl.") or "unsloth" in _m.lower():
        del sys.modules[_m]
sys.meta_path = [f for f in sys.meta_path
                 if "unsloth" not in type(f).__module__.lower()]

from transformers import Trainer, TrainingArguments

# vanilla transformers.Trainer (NOT trl.SFTTrainer). SFTTrainer + its dataset-prep
# path (skip_prepare_dataset / its own collator) could drop the masked `labels` ->
# nll over an empty mask -> nan -> the old `logits.sum()*0.0` guard turned that into
# a clean 0.0 forever. Plain Trainer passes input_ids/labels to compute_loss verbatim.


class RaftTrainer(Trainer):
    """Plain mean-NLL SFT on the kept correct rollouts (self-distillation).

    Fused cross_entropy(reduction="none") -> no fp32 full-vocab log_softmax spike.
    fp32 reduction; non-finite logits sanitised+recomputed (real grad, not a dead
    zero); loud counters so a silent 0.0 can never recur unnoticed.
    """
    def __init__(self, *a, **k):
        super().__init__(*a, **k)
        self._c = {"nonfinite": 0, "dead": 0, "dbg": 0}

    def compute_loss(self, model, inputs, return_outputs=False, **kw):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits
        sl = logits[:, :-1, :]
        lab = labels[:, 1:].to(sl.device)
        B, T, V = sl.shape
        nll = F.cross_entropy(sl.reshape(-1, V), lab.reshape(-1),
                              ignore_index=-100, reduction="none").view(B, T)
        if not torch.isfinite(nll).all():
            self._c["nonfinite"] += 1
            sl2 = torch.nan_to_num(sl, nan=0.0, posinf=30.0, neginf=-30.0)
            nll = F.cross_entropy(sl2.reshape(-1, V), lab.reshape(-1),
                                  ignore_index=-100, reduction="none").view(B, T)
            nll = torch.nan_to_num(nll, nan=0.0, posinf=30.0, neginf=0.0)
        mask = (lab != -100)
        loss = nll.float()[mask].mean()

        if self._c["dbg"] < 3:
            self._c["dbg"] += 1
            print(f"[loss-dbg step~{self.state.global_step}] loss={float(loss):.4f} "
                  f"unmasked={int(mask.sum())} nonfinite={self._c['nonfinite']}")

        if not torch.isfinite(loss):
            self._c["dead"] += 1
            if self._c["dead"] <= 5 or self._c["dead"] % 50 == 0:
                print(f"[loss-WARN] non-finite loss step {self.state.global_step} "
                      f"(dead={self._c['dead']}); zeroing. Lower TRAIN_MAX_LEN/batch if frequent.")
            loss = (logits.float().sum() * 0.0).requires_grad_(True)
        return (loss, outputs) if return_outputs else loss


print("RaftTrainer defined (vanilla transformers.Trainer).")
```

## Iterate (optional, next session)

RAFT is a loop. After the real run:
1. Upload `outputs/raft_adapter` as a Kaggle dataset, point `SFT_ADAPTER_DIR` at it.
2. Delete `raft_rollouts.jsonl` so the stronger model re-samples.
3. Re-run. Prompts that were all-wrong last round may now become mixed (curriculum
   moves up). 2-3 rounds typically.

Watch: corpus size should grow across rounds (more prompts solvable), and the
`all_wrong` count should shrink.
